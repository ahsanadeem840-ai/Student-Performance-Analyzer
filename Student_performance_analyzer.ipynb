```python
data = pd.read_csv("students.csv")
data.head()
```

```python
print(data.head())
```

```python
print(data.shape)
```

```python
print(data.info())
```

```python
data["Total"] = data["Math"] + data["English"] + data["Science"] + data["Computer"]

print(data)
```

```python
data["Percentage"] = (data["Total"]/400)*100
print(data)
```

```python
highest_marks = data.loc[data["Percentage"].idxmax()]
print("Highest Performance Student")
print(highest_marks)
```

```python
highest_marks = data.loc[data["Percentage"].idxmin()]
print("Highest Performance Student")
print(highest_marks)
```

```python
class_average = data["Percentage"].mean()
print("Class Average")
print(class_average)
```

```python
above_80 = data[data["Percentage"]>80]
print("Performance greater then 80")
print(above_80)
```

```python
percentages = np.array(data["Percentage"])

print("Percentages:")
print(percentages)
```

```python
print("Maximum Percentage:", np.max(percentages))
print("Minimum Percentage:", np.min(percentages))
print("Average Percentage:", np.mean(percentages))
```

```python
import matplotlib as plt
data = pd.read_csv("students.csv")
data.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(data["English"], data["Computer"])

plt.title("Student Performance")
plt.xlabel("Students")
plt.ylabel("Percentage")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(data=data, x="Math", y="Science")

plt.title("Student Percentage Comparison")
plt.show()
```

```python
subject_average = data[["Math", "English", "Science", "Computer"]].mean()

print(subject_average)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(x=subject_average.index, y=subject_average.values)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```
